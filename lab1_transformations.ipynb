```python
def plot_off(vertices, faces):
    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(111)

    mesh = PolyCollection([vertices[face] for face in faces],
                          alpha=0.3, edgecolors='k')
    ax.add_collection(mesh)
    ax.scatter(vertices[:, 0], vertices[:, 1], s=10, c="pink", zorder=5)

    ax.set_xlabel("X")
    ax.set_ylabel("Y")

    ax.autoscale_view()
    ax.set_aspect('equal')

    plt.show()
```

### task 1

```python
def stretch(X: np.ndarray, a: float, b: float):
    T = np.array([[a, 0], [0, b]])
    return X.copy() @ T
stretch_lynx = stretch(lynx, 1.5, 0.7)
print(stretch_lynx)
plot_off(stretch_lynx, faces)
```

```python
def shear(X: np.ndarray, a: float, b: float):
    T = np.array([[1, b], [a, 1]])
    return X.copy() @ T
shear_lynx = shear(lynx, 0.5, 0)
print(shear_lynx)
plot_off(shear_lynx, faces)
```

```python
def reflection(X: np.ndarray, a: float, b: float):
    T = np.array([[a**2 - b**2, 2*a*b], [2*a*b, b**2-a**2]])
    return X.copy() @ (T/(a**2+b**2))
reflection_lynx = reflection(lynx, 1, 1)
plot_off(reflection_lynx, faces)
print(reflection_lynx)
```

```python
def rotation(X: np.ndarray, degrees):
    rad = np.radians(degrees)
    T = np.array([[np.cos(rad), np.sin(rad)], [-np.sin(rad), np.cos(rad)]])
    return X.copy() @ T
rotation_lynx = rotation(lynx, 45)
plot_off(rotation_lynx, faces)
print(rotation_lynx)
```
