### task 1

```python
def stretch(X: np.ndarray, a: float, b: float):
    T = np.array([[a, 0], [0, b]])
    return X.copy() @ T
stretch_lynx = stretch(lynx, 1.5, 0.7)
print(stretch_lynx)
plot_off(stretch_lynx, faces)
```

```python
def shear(X: np.ndarray, a: float, b: float):
    T = np.array([[1, b], [a, 1]])
    return X.copy() @ T
shear_lynx = shear(lynx, 0.5, 0)
print(shear_lynx)
plot_off(shear_lynx, faces)
```

```python
def reflection(X: np.ndarray, a: float, b: float):
    T = np.array([[a**2 - b**2, 2*a*b], [2*a*b, b**2-a**2]])
    return X.copy() @ (T/(a**2+b**2))
reflection_lynx = reflection(lynx, 1, 1)
plot_off(reflection_lynx, faces)
print(reflection_lynx)
```

```python
def rotation(X: np.ndarray, degrees):
    rad = np.radians(degrees)
    T = np.array([[np.cos(rad), np.sin(rad)], [-np.sin(rad), np.cos(rad)]])
    return X.copy() @ T
rotation_lynx = rotation(lynx, 45)
plot_off(rotation_lynx, faces)
print(rotation_lynx)
```

```python
def read_off(filename: str):
    with open (filename, 'r') as f:
        # Перевіряємо, чи перший рядок починається з OFF
        if 'OFF' != f.readline().strip():
            raise ValueError( 'Not a valid OFF header')
        # Зчитуємо кількість вершин, граней та ребер (трете значення частоі гнорується)
        n_verts, n_faces, _ = map(int, f.readline().strip().split())
        # Зчитуємо координати всіх вершин (х, у, z)
        verts = [list(map(float, f.readline().strip().split())) for _ in range(n_verts)]
        # Зчитуємо грані: перше число у рядку - кількість вершин грані (ігноруємо),далі індекси
        faces = [list(map(int, f.readline().strip().split()[1:])) for _ in range(n_faces)]
        # Повертаємо вершини у вигляді масиву NumPy та список граней
    return np.array(verts), faces
```

```python
def plot_off_3d(vertices, faces):
    fig = plt.figure(figsize=(8, 8))
    ax = fig.add_subplot(111, projection='3d')

    mesh = Poly3DCollection([vertices[face] for face in faces], alpha=0.3, edgecolor='k')

    ax.add_collection3d(mesh)
    # Додаємо вершини як червоні точки
    ax.scatter(vertices[:, 0], vertices[:, 1], vertices [:, 2], s=2, c= 'r')
    # Підписуємо осі
    ax.set_xlabel("X")
    ax.set_ylabel("Y")
    ax.set_zlabel("Z")
    # Автоматично масштабуємо сцену під модель
    ax.auto_scale_xyz(vertices[:, 0], vertices[:, 1], vertices[:, 2])
     # показуємо результат
    #plt.close(fig)
    # Виклик функції для побудови OFF-моделі
```

```python
vertices, faces = read_off("/Users/maria/.cache/kagglehub/datasets/balraj98/modelnet40-princeton-3d-object-dataset/versions/1/ModelNet40/airplane/test/airplane_0627.off")
print(vertices.shape)
print (faces [:3])
print(vertices)
plot_off_3d(vertices, faces)
```

```python
def rotate_xy(X: np.ndarray, degrees):
    rad = np.radians(degrees)
    T = np.array([[np.cos(rad), np.sin(rad), 0],
                  [-np.sin(rad), np.cos(rad), 0 ],
                  [0, 0, 1]])
    return X.copy() @ T
rotated_xy = rotate_xy(vertices, 90)
plot_off_3d(rotated_xy, faces)
print(rotated_xy)
```

```python
def rotate_yz(X: np.ndarray, degrees):
    rad = np.radians(degrees)
    T = np.array([[1,0, 0],
                  [0, np.cos(rad), np.sin(rad)],
                  [0, -np.sin(rad), np.cos(rad)]])
    return X.copy() @ T


rotated_yz = rotate_yz(vertices, 90)
plot_off_3d(rotated_yz, faces)
print(rotated_yz)
```

```python
def rotate_xz(X: np.ndarray, degrees):
    rad = np.radians(degrees)
    T = np.array([[np.cos(rad), 0, np.sin(rad)],
                  [0, 1, 0],
                  [-np.sin(rad), 0, np.cos(rad)]])
    return X.copy() @ T
rotated_xz = rotate_xz(vertices, 90)
plot_off_3d(rotated_xz, faces)
print(rotated_xz)
```

```python
combination = rotate_xz(rotate_yz(rotate_xy(vertices, 90), 90), 90)
plot_off_3d(combination, faces)
print(combination)
```
